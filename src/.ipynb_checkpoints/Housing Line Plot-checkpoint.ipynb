```python
#!/usr/bin/env python
# coding: utf-8

import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np

# Read Zillow data
zillow = pd.read_csv('../Data/zillow county.csv')

# Select and convert datetime columns
column_names_to_convert = zillow.columns[9:]
new_column_names = pd.to_datetime(column_names_to_convert)
zillow.columns = list(zillow.columns[:9]) + list(new_column_names)

# Drop columns outside the desired year range (2017-2019)
for column in new_column_names:
    column_year = column.year
    if not (2017 <= column_year <= 2019):
        zillow = zillow.drop(columns=[column])

# Make county FIPS column
zillow['CountyFIPS'] = (
    zillow['StateCodeFIPS'].astype(str).str.zfill(2) + 
    zillow['MunicipalCodeFIPS'].astype(str).str.zfill(3)
)

# Read tariff data
tariffs = pd.read_csv('../Data/ustariffs-by-county.csv')

# Clean tariff data
tariffs_clean = tariffs.drop(columns=['Unnamed: 0','emplvl_2017', 'fips', 'total_employment', 'int_area_fips',
       '2017_income', '2017_population', 'total_imp_pc', 'china_imp_pc'])

# Filter for years and clean area_fips
tariffs_clean['time'] = pd.to_datetime(tariffs_clean['time'])
tariffs_clean = tariffs_clean[(tariffs_clean['time'].dt.year >= 2017) & (tariffs_clean['time'].dt.year <= 2019)]
tariffs_clean['area_fips'] = tariffs_clean['area_fips'].astype(str).str.zfill(5)
tariffs_clean = tariffs_clean[~tariffs_clean['area_fips'].astype(str).str.endswith('999')]
tariffs_clean = tariffs_clean[tariffs_clean['time'] != '2017-01-01 00:00:00']
if 'state' in tariffs_clean.columns:
    tariffs_clean = tariffs_clean.drop('state', axis=1)
tariffs_clean = tariffs_clean.reset_index(drop=True)

# Pivot tariff data and calculate change
tariffs_clean_pivoted = tariffs_clean.pivot(index='area_fips', columns='time', values='tariff')
tariffs_clean_pivoted = tariffs_clean_pivoted.reset_index()
tariffs_clean_pivoted['change'] = tariffs_clean_pivoted.iloc[:,-1] - tariffs_clean_pivoted.iloc[:,1]
tariffs_clean_pivoted = tariffs_clean_pivoted.sort_values(by=['change'], ascending=False)
tariffs_clean_pivoted = tariffs_clean_pivoted.reset_index(drop=True)
tariffs_clean_pivoted = tariffs_clean_pivoted.dropna()

# Create quartiles based on tariff changes
tariffs_clean_pivoted['quartile'] = pd.qcut(tariffs_clean_pivoted['change'], q=4, labels=['Q1', 'Q2', 'Q3', 'Q4'])

# Get all quartiles
q1 = tariffs_clean_pivoted[tariffs_clean_pivoted['quartile'] == 'Q1']
q2 = tariffs_clean_pivoted[tariffs_clean_pivoted['quartile'] == 'Q2']
q3 = tariffs_clean_pivoted[tariffs_clean_pivoted['quartile'] == 'Q3']
q4 = tariffs_clean_pivoted[tariffs_clean_pivoted['quartile'] == 'Q4']

# Clean data columns to drop
drop_columns = ['Metro','StateCodeFIPS','MunicipalCodeFIPS','State','StateName','RegionType','RegionName','SizeRank','RegionID','CountyFIPS']

# Function to create quartile comparison plot
# Function to create quartile comparison plot with larger fonts
# Function to create quartile comparison plot with larger fonts
# Function to create quartile comparison plot with larger fonts
def create_quartile_plot(q_high, q_low, q_high_name, q_low_name, title_suffix, filename):
    # Filter Zillow data for the quartiles
    zillow_high_tariffs = zillow[zillow['CountyFIPS'].isin(q_high['area_fips'])]
    zillow_low_tariffs = zillow[zillow['CountyFIPS'].isin(q_low['area_fips'])]
    
    # Clean data and calculate means
    zillow_high_clean = zillow_high_tariffs.drop(columns=drop_columns)
    zillow_low_clean = zillow_low_tariffs.drop(columns=drop_columns)
    
    # Calculate log means for both difference and individual quartiles
    zillow_high_log_mean = pd.DataFrame([np.log(zillow_high_clean.mean())])
    zillow_low_log_mean = pd.DataFrame([np.log(zillow_low_clean.mean())])
    diff = zillow_high_log_mean - zillow_low_log_mean
    
    # Calculate averages before and after the split date
    split_date = pd.Timestamp('2018-02-28')
    before_mask = diff.columns <= split_date
    after_mask = diff.columns > split_date
    before_mean = diff.iloc[0, before_mask].mean()
    after_mean = diff.iloc[0, after_mask].mean()
    
    # Create the plot with dual y-axes
    fig, ax1 = plt.subplots(figsize=(15, 8))
    
    # Convert dates to month format and shift them forward by one month for x-axis
    x_dates = [(d + pd.DateOffset(months=1)).strftime('%Y-%m') for d in diff.columns]
    
    # Main line plot (price difference) on left y-axis
    ax1.plot(x_dates, diff.iloc[0], 
             marker='o', 
             linewidth=2, 
             markersize=6,
             color='#2E86C1',
             linestyle='-',
             markerfacecolor='white',
             markeredgewidth=2,
             label=f'Price Difference ({q_high_name}-{q_low_name})')
    
    # Create second y-axis for log average housing prices
    ax2 = ax1.twinx()
    
    # Plot log mean prices for both quartiles on right y-axis
    ax2.plot(x_dates, zillow_low_log_mean.iloc[0].values, 
             marker='s', 
             linewidth=2, 
             markersize=5,
             color='orange',
             linestyle='--',
             alpha=0.8,
             label=f'{q_low_name} Log Avg Prices')
    
    ax2.plot(x_dates, zillow_high_log_mean.iloc[0].values, 
             marker='^', 
             linewidth=2, 
             markersize=5,
             color='purple',
             linestyle='--',
             alpha=0.8,
             label=f'{q_high_name} Log Avg Prices')
    
    # Add horizontal mean lines for price difference (left axis)
    ax1.axhline(y=before_mean, color='r', linestyle='--', alpha=0.7, 
               xmin=0, xmax=len(before_mask[before_mask])/len(diff.columns),
               label=f'Pre-Tariff Mean: {before_mean:.2f}')
    ax1.axhline(y=after_mean, color='g', linestyle='--', alpha=0.7,
               xmin=len(before_mask[before_mask])/len(diff.columns), xmax=1,
               label=f'Post-Tariff Mean: {after_mean:.2f}')
    
    # Get y-axis limits for consistent positioning
    y_min, y_max = ax1.get_ylim()
    y_range = y_max - y_min
    
    # Add line for steel and aluminum tariff announcement
    steel_date = pd.Timestamp('2018-01-31')
    shifted_steel_date = (steel_date + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax1.axvline(x=shifted_steel_date, color='blue', linestyle=':', alpha=0.5)
    ax1.text(shifted_steel_date, y_max - 0.05 * y_range, 
             'US government increases\ntariffs on steel and aluminum',
             verticalalignment='top',
             horizontalalignment='left',
             color='blue',
             fontsize=14)
    
    # Add line for releasing list
    list_date = pd.Timestamp('2018-03-31')
    shifted_list_date = (list_date + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax1.axvline(x=shifted_list_date, color='blue', linestyle=':', alpha=0.5)
    ax1.text(shifted_list_date, y_max - 0.15 * y_range, 
             'US government releases\n50B list of Chinese Products',
             verticalalignment='top',
             horizontalalignment='left',
             color='blue',
             fontsize=14)
    
    # Add line for implementation date
    tariff_date = pd.Timestamp('2018-05-31')
    shifted_tariff_date = (tariff_date + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax1.axvline(x=shifted_tariff_date, color='blue', linestyle=':', alpha=0.5)
    ax1.text(shifted_tariff_date, y_min + 0.15 * y_range, 
             'China and US implement\n34B worth of Tariffs',
             verticalalignment='bottom',
             horizontalalignment='right',
             color='blue',
             fontsize=14)
    
    tariff_date2 = pd.Timestamp('2018-07-31')
    shifted_tariff_date2 = (tariff_date2 + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax1.axvline(x=shifted_tariff_date2, color='blue', linestyle=':', alpha=0.5)
    ax1.text(shifted_tariff_date2, y_min + 0.05 * y_range, 
             'China and US implement\nremaining 17B worth of Tariffs',
             verticalalignment='bottom',
             horizontalalignment='left',
             color='blue',
             fontsize=14)
    
    # Improve grid and formatting
    ax1.grid(True, linestyle='--', alpha=0.7, axis='y')
    ax1.set_title(f'Housing Price Changes Over Time ({title_suffix})', fontsize=18, pad=15)  # Increased from 14 to 18
    ax1.set_xlabel('Time Period', fontsize=16)  # Increased from 12 to 16
    ax1.set_ylabel('(Log) Price Difference', fontsize=16, color='#2E86C1')  # Increased from 12 to 16
    ax2.set_ylabel('Log Average Housing Price', fontsize=16, color='black')  # Increased from 12 to 16
    
    # Format x-axis with larger tick labels
    ax1.tick_params(axis='x', rotation=45, labelsize=14)  # Added labelsize=14
    ax1.tick_params(axis='y', labelcolor='#2E86C1', labelsize=14)  # Added labelsize=14
    ax2.tick_params(axis='y', labelcolor='black', labelsize=14)  # Added labelsize=14
    plt.setp(ax1.xaxis.get_majorticklabels(), ha='right')
    
    # Combine legends from both axes with larger font
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, 
               labels1 + labels2,
               loc='upper left',
               fontsize=14)  # Increased legend font size
    
    # Remove top and right borders for both axes
    ax1.spines['top'].set_visible(False)
    ax2.spines['top'].set_visible(False)
    
    # Clean up layout
    fig.tight_layout()
    
    # Save the figure
    plt.savefig(f'../images/{filename}', 
                dpi=300,
                bbox_inches='tight',
                facecolor='white',
                format='png')
    
    # Show the plot
    plt.show()
    
    print(f"Plot saved as: images/{filename}")

# Generate all four quartile comparison plots
print("Generating Q1 vs Q4 comparison...")
create_quartile_plot(q4, q1, 'Q4', 'Q1', 'Q1 vs Q4', 'housing_price_changes_q1_q4.png')

print("\nGenerating Q2 vs Q3 comparison...")
create_quartile_plot(q3, q2, 'Q3', 'Q2', 'Q2 vs Q3', 'housing_price_changes_q2_q3.png')

print("\nGenerating Q1 vs Q2 comparison...")
create_quartile_plot(q2, q1, 'Q2', 'Q1', 'Q1 vs Q2', 'housing_price_changes_q1_q2.png')

print("\nGenerating Q1 vs Q3 comparison...")
create_quartile_plot(q3, q1, 'Q3', 'Q1', 'Q1 vs Q3', 'housing_price_changes_q1_q3.png')

print("\nAll plots generated successfully!")
```

```python
# Create a plot showing all 4 quartiles and overall mean housing prices
def create_all_quartiles_plot():
    # Filter Zillow data for each quartile
    zillow_q1 = zillow[zillow['CountyFIPS'].isin(q1['area_fips'])]
    zillow_q2 = zillow[zillow['CountyFIPS'].isin(q2['area_fips'])]
    zillow_q3 = zillow[zillow['CountyFIPS'].isin(q3['area_fips'])]
    zillow_q4 = zillow[zillow['CountyFIPS'].isin(q4['area_fips'])]
    
    # Clean data for each quartile
    zillow_q1_clean = zillow_q1.drop(columns=drop_columns)
    zillow_q2_clean = zillow_q2.drop(columns=drop_columns)
    zillow_q3_clean = zillow_q3.drop(columns=drop_columns)
    zillow_q4_clean = zillow_q4.drop(columns=drop_columns)
    
    # Calculate log means for each quartile
    zillow_q1_log_mean = np.log(zillow_q1_clean.mean())
    zillow_q2_log_mean = np.log(zillow_q2_clean.mean())
    zillow_q3_log_mean = np.log(zillow_q3_clean.mean())
    zillow_q4_log_mean = np.log(zillow_q4_clean.mean())
    
    # Calculate overall mean (all counties combined)
    all_counties_clean = zillow.drop(columns=drop_columns)
    overall_log_mean = np.log(all_counties_clean.mean())
    
    # Convert dates to month format and shift them forward by one month for x-axis
    x_dates = [(d + pd.DateOffset(months=1)).strftime('%Y-%m') for d in zillow_q1_log_mean.index]
    
    # Create the plot
    fig, ax = plt.subplots(figsize=(15, 8))
    
    # Define colors and styles for each quartile
    colors = ['#E74C3C', '#F39C12', '#2ECC71', '#3498DB']  # Red, Orange, Green, Blue
    markers = ['o', 's', '^', 'D']  # Circle, Square, Triangle, Diamond
    
    # Plot each quartile
    ax.plot(x_dates, zillow_q1_log_mean.values, 
            marker=markers[0], linewidth=2.5, markersize=6,
            color=colors[0], linestyle='-', markerfacecolor='white',
            markeredgewidth=2, label='Q1 (Lowest Tariff Change)',
            alpha=0.9)
    
    ax.plot(x_dates, zillow_q2_log_mean.values, 
            marker=markers[1], linewidth=2.5, markersize=6,
            color=colors[1], linestyle='-', markerfacecolor='white',
            markeredgewidth=2, label='Q2',
            alpha=0.9)
    
    ax.plot(x_dates, zillow_q3_log_mean.values, 
            marker=markers[2], linewidth=2.5, markersize=6,
            color=colors[2], linestyle='-', markerfacecolor='white',
            markeredgewidth=2, label='Q3',
            alpha=0.9)
    
    ax.plot(x_dates, zillow_q4_log_mean.values, 
            marker=markers[3], linewidth=2.5, markersize=6,
            color=colors[3], linestyle='-', markerfacecolor='white',
            markeredgewidth=2, label='Q4 (Highest Tariff Change)',
            alpha=0.9)
    
    # Plot overall mean
    ax.plot(x_dates, overall_log_mean.values, 
            marker='*', linewidth=3, markersize=8,
            color='black', linestyle='--', markerfacecolor='gold',
            markeredgewidth=2, label='Overall Mean (All Counties)',
            alpha=0.8)
    
    # Add vertical lines for tariff events
    steel_date = pd.Timestamp('2018-01-31')
    shifted_steel_date = (steel_date + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax.axvline(x=shifted_steel_date, color='gray', linestyle=':', alpha=0.6, linewidth=2)
    
    list_date = pd.Timestamp('2018-03-31')
    shifted_list_date = (list_date + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax.axvline(x=shifted_list_date, color='gray', linestyle=':', alpha=0.6, linewidth=2)
    
    tariff_date = pd.Timestamp('2018-05-31')
    shifted_tariff_date = (tariff_date + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax.axvline(x=shifted_tariff_date, color='gray', linestyle=':', alpha=0.6, linewidth=2)
    
    tariff_date2 = pd.Timestamp('2018-07-31')
    shifted_tariff_date2 = (tariff_date2 + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax.axvline(x=shifted_tariff_date2, color='gray', linestyle=':', alpha=0.6, linewidth=2)
    
    # Get y-axis limits for text positioning
    y_min, y_max = ax.get_ylim()
    y_range = y_max - y_min
    
    # Add annotations for key events
    ax.text(shifted_steel_date, y_max - 0.05 * y_range, 
            'Steel/Aluminum\nTariffs',
            verticalalignment='top',
            horizontalalignment='center',
            color='gray',
            fontsize=12,
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
    
    ax.text(shifted_list_date, y_max - 0.15 * y_range, 
            '$50B List\nReleased',
            verticalalignment='top',
            horizontalalignment='center',
            color='gray',
            fontsize=12,
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
    
    ax.text(shifted_tariff_date, y_min + 0.15 * y_range, 
            '$34B Tariffs\nImplemented',
            verticalalignment='bottom',
            horizontalalignment='center',
            color='gray',
            fontsize=12,
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
    
    ax.text(shifted_tariff_date2, y_min + 0.05 * y_range, 
            '$17B Tariffs\nImplemented',
            verticalalignment='bottom',
            horizontalalignment='center',
            color='gray',
            fontsize=12,
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
    
    # Formatting
    ax.grid(True, linestyle='--', alpha=0.4, axis='y')
    ax.set_title('Log Housing Prices by Tariff Change Quartiles Over Time', fontsize=18, pad=20)
    ax.set_xlabel('Time Period', fontsize=16)
    ax.set_ylabel('Log Average Housing Price', fontsize=16)
    
    # Format x-axis
    ax.tick_params(axis='x', rotation=45, labelsize=14)
    ax.tick_params(axis='y', labelsize=14)
    plt.setp(ax.xaxis.get_majorticklabels(), ha='right')
    
    # Add legend
    ax.legend(loc='upper left', fontsize=13, frameon=True, 
              fancybox=True, shadow=True, framealpha=0.9)
    
    # Remove top and right borders
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    
    # Add subtle background color
    ax.set_facecolor('#FAFAFA')
    
    # Clean up layout
    fig.tight_layout()
    
    # Save the figure
    plt.savefig('../images/all_quartiles_housing_prices.png', 
                dpi=300,
                bbox_inches='tight',
                facecolor='white',
                format='png')
    
    # Show the plot
    plt.show()
    
    print("All quartiles plot saved as: images/all_quartiles_housing_prices.png")
    
    # Print some summary statistics
    print("\n=== SUMMARY STATISTICS ===")
    
    # Calculate pre and post tariff means for each quartile
    split_date = pd.Timestamp('2018-02-28')
    
    for i, (q_data, q_name) in enumerate([(zillow_q1_log_mean, 'Q1'), 
                                          (zillow_q2_log_mean, 'Q2'), 
                                          (zillow_q3_log_mean, 'Q3'), 
                                          (zillow_q4_log_mean, 'Q4')]):
        before_mask = q_data.index <= split_date
        after_mask = q_data.index > split_date
        
        before_mean = q_data[before_mask].mean()
        after_mean = q_data[after_mask].mean()
        change = after_mean - before_mean
        
        print(f"{q_name}: Pre-tariff mean = {before_mean:.4f}, Post-tariff mean = {after_mean:.4f}, Change = {change:+.4f}")
    
    # Overall statistics
    before_mask = overall_log_mean.index <= split_date
    after_mask = overall_log_mean.index > split_date
    
    overall_before_mean = overall_log_mean[before_mask].mean()
    overall_after_mean = overall_log_mean[after_mask].mean()
    overall_change = overall_after_mean - overall_before_mean
    
    print(f"Overall: Pre-tariff mean = {overall_before_mean:.4f}, Post-tariff mean = {overall_after_mean:.4f}, Change = {overall_change:+.4f}")

# Generate the plot
print("Generating all quartiles comparison plot...")
create_all_quartiles_plot()
print("Plot generated successfully!")
```

```python
# Create a plot showing normalized quartile deviations from starting point
def create_normalized_quartiles_plot():
    # Filter Zillow data for each quartile
    zillow_q1 = zillow[zillow['CountyFIPS'].isin(q1['area_fips'])]
    zillow_q2 = zillow[zillow['CountyFIPS'].isin(q2['area_fips'])]
    zillow_q3 = zillow[zillow['CountyFIPS'].isin(q3['area_fips'])]
    zillow_q4 = zillow[zillow['CountyFIPS'].isin(q4['area_fips'])]
    
    # Clean data for each quartile
    zillow_q1_clean = zillow_q1.drop(columns=drop_columns)
    zillow_q2_clean = zillow_q2.drop(columns=drop_columns)
    zillow_q3_clean = zillow_q3.drop(columns=drop_columns)
    zillow_q4_clean = zillow_q4.drop(columns=drop_columns)
    
    # Calculate log means for each quartile
    zillow_q1_log_mean = np.log(zillow_q1_clean.mean())
    zillow_q2_log_mean = np.log(zillow_q2_clean.mean())
    zillow_q3_log_mean = np.log(zillow_q3_clean.mean())
    zillow_q4_log_mean = np.log(zillow_q4_clean.mean())
    
    # Get the starting date (first observation) for normalization
    start_date = zillow_q1_log_mean.index[0]
    
    # Normalize each series to start at 0 (deviation from starting point)
    q1_normalized = zillow_q1_log_mean - zillow_q1_log_mean.iloc[0]
    q2_normalized = zillow_q2_log_mean - zillow_q2_log_mean.iloc[0]
    q3_normalized = zillow_q3_log_mean - zillow_q3_log_mean.iloc[0]
    q4_normalized = zillow_q4_log_mean - zillow_q4_log_mean.iloc[0]
    
    # Convert dates to month format and shift them forward by one month for x-axis
    x_dates = [(d + pd.DateOffset(months=1)).strftime('%Y-%m') for d in q1_normalized.index]
    
    # Create the plot
    fig, ax = plt.subplots(figsize=(16, 10))
    
    # Define colors and styles for quartiles only
    colors = ['#E74C3C', '#F39C12', '#2ECC71', '#3498DB']  # Red, Orange, Green, Blue
    markers = ['o', 's', '^', 'D']  # Circle, Square, Triangle, Diamond
    linewidths = [2.5, 2.5, 2.5, 2.5]
    alphas = [0.9, 0.9, 0.9, 0.9]
    
    # Plot each normalized quartile
    ax.plot(x_dates, q1_normalized.values, 
            marker=markers[0], linewidth=linewidths[0], markersize=6,
            color=colors[0], linestyle='-', markerfacecolor='white',
            markeredgewidth=2, label='Q1 (Lowest Tariff Change)',
            alpha=alphas[0])
    
    ax.plot(x_dates, q2_normalized.values, 
            marker=markers[1], linewidth=linewidths[1], markersize=6,
            color=colors[1], linestyle='-', markerfacecolor='white',
            markeredgewidth=2, label='Q2',
            alpha=alphas[1])
    
    ax.plot(x_dates, q3_normalized.values, 
            marker=markers[2], linewidth=linewidths[2], markersize=6,
            color=colors[2], linestyle='-', markerfacecolor='white',
            markeredgewidth=2, label='Q3',
            alpha=alphas[2])
    
    ax.plot(x_dates, q4_normalized.values, 
            marker=markers[3], linewidth=linewidths[3], markersize=6,
            color=colors[3], linestyle='-', markerfacecolor='white',
            markeredgewidth=2, label='Q4 (Highest Tariff Change)',
            alpha=alphas[3])
    
    # Add horizontal reference line at zero (starting point)
    ax.axhline(y=0, color='black', linestyle='-', linewidth=1, alpha=0.5,
               label='Starting Point (Normalized to 0)')
    
    # Add vertical lines for tariff events
    steel_date = pd.Timestamp('2018-01-31')
    shifted_steel_date = (steel_date + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax.axvline(x=shifted_steel_date, color='gray', linestyle=':', alpha=0.6, linewidth=2)
    
    list_date = pd.Timestamp('2018-03-31')
    shifted_list_date = (list_date + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax.axvline(x=shifted_list_date, color='gray', linestyle=':', alpha=0.6, linewidth=2)
    
    tariff_date = pd.Timestamp('2018-05-31')
    shifted_tariff_date = (tariff_date + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax.axvline(x=shifted_tariff_date, color='gray', linestyle=':', alpha=0.6, linewidth=2)
    
    tariff_date2 = pd.Timestamp('2018-07-31')
    shifted_tariff_date2 = (tariff_date2 + pd.DateOffset(months=1)).strftime('%Y-%m')
    ax.axvline(x=shifted_tariff_date2, color='gray', linestyle=':', alpha=0.6, linewidth=2)
    
    # Get y-axis limits for text positioning
    y_min, y_max = ax.get_ylim()
    y_range = y_max - y_min
    
    # Add annotations for key events
    ax.text(shifted_steel_date, y_max - 0.05 * y_range, 
            'Steel/Aluminum\nTariffs',
            verticalalignment='top',
            horizontalalignment='center',
            color='gray',
            fontsize=12,
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
    
    ax.text(shifted_list_date, y_max - 0.15 * y_range, 
            '$50B List\nReleased',
            verticalalignment='top',
            horizontalalignment='center',
            color='gray',
            fontsize=12,
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
    
    ax.text(shifted_tariff_date, y_min + 0.15 * y_range, 
            '$34B Tariffs\nImplemented',
            verticalalignment='bottom',
            horizontalalignment='center',
            color='gray',
            fontsize=12,
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
    
    ax.text(shifted_tariff_date2, y_min + 0.05 * y_range, 
            '$17B Tariffs\nImplemented',
            verticalalignment='bottom',
            horizontalalignment='center',
            color='gray',
            fontsize=12,
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))
    
    # Formatting
    ax.grid(True, linestyle='--', alpha=0.4, axis='y')
    ax.set_title(f'Normalized Housing Price Deviations by Tariff Change Quartiles\n(All Series Start at 0 on {start_date.strftime("%Y-%m")})', 
                 fontsize=18, pad=20)
    ax.set_xlabel('Time Period', fontsize=16)
    ax.set_ylabel('Log Price Change from Starting Point', fontsize=16)
    
    # Format x-axis
    ax.tick_params(axis='x', rotation=45, labelsize=14)
    ax.tick_params(axis='y', labelsize=14)
    plt.setp(ax.xaxis.get_majorticklabels(), ha='right')
    
    # Add legend
    ax.legend(loc='upper left', fontsize=13, frameon=True, 
              fancybox=True, shadow=True, framealpha=0.9)
    
    # Remove top and right borders
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    
    # Add subtle background color
    ax.set_facecolor('#FAFAFA')
    
    # Clean up layout
    fig.tight_layout()
    
    # Save the figure
    plt.savefig('../images/normalized_quartiles_housing_deviations.png', 
                dpi=300,
                bbox_inches='tight',
                facecolor='white',
                format='png')
    
    # Show the plot
    plt.show()
    
    print("Normalized quartiles deviation plot saved as: images/normalized_quartiles_housing_deviations.png")
    
    # Print detailed analysis
    print(f"\n=== NORMALIZED DEVIATION ANALYSIS ===")
    print(f"Starting Date: {start_date.strftime('%Y-%m-%d')}")
    print(f"All series normalized to 0 at starting point")
    
    # Calculate final deviations
    final_deviations = {
        'Q1': q1_normalized.iloc[-1],
        'Q2': q2_normalized.iloc[-1], 
        'Q3': q3_normalized.iloc[-1],
        'Q4': q4_normalized.iloc[-1]
    }
    
    print(f"\n=== FINAL DEVIATIONS FROM START ===")
    for quartile, deviation in final_deviations.items():
        print(f"{quartile}: {deviation:+.4f} log points ({(np.exp(deviation)-1)*100:+.2f}% price change)")
    
    # Calculate period-specific changes
    split_date1 = pd.Timestamp('2018-01-31')
    split_date2 = pd.Timestamp('2018-05-31')
    
    # Find the closest dates in our data
    pre_end_idx = q1_normalized.index.get_loc(split_date1, method='nearest')
    announcement_end_idx = q1_normalized.index.get_loc(split_date2, method='nearest')
    
    print(f"\n=== PERIOD-SPECIFIC CHANGES ===")
    
    data_series = {
        'Q1': q1_normalized,
        'Q2': q2_normalized,
        'Q3': q3_normalized, 
        'Q4': q4_normalized
    }
    
    for name, series in data_series.items():
        pre_change = series.iloc[pre_end_idx] - series.iloc[0]
        announcement_change = series.iloc[announcement_end_idx] - series.iloc[pre_end_idx]
        post_change = series.iloc[-1] - series.iloc[announcement_end_idx]
        
        print(f"\n{name}:")
        print(f"  Pre-Tariff Change: {pre_change:+.4f} log points")
        print(f"  Announcement Period Change: {announcement_change:+.4f} log points")
        print(f"  Post-Implementation Change: {post_change:+.4f} log points")
    
    # Calculate quartile-to-quartile relative performance
    print(f"\n=== QUARTILE-TO-QUARTILE COMPARISON ===")
    print(f"Q4 vs Q1: {data_series['Q4'].iloc[-1] - data_series['Q1'].iloc[-1]:+.4f} log points")
    print(f"Q3 vs Q2: {data_series['Q3'].iloc[-1] - data_series['Q2'].iloc[-1]:+.4f} log points")
    print(f"Q4 vs Q2: {data_series['Q4'].iloc[-1] - data_series['Q2'].iloc[-1]:+.4f} log points")
    print(f"Q3 vs Q1: {data_series['Q3'].iloc[-1] - data_series['Q1'].iloc[-1]:+.4f} log points")
    
    # Calculate maximum and minimum deviations for each quartile
    print(f"\n=== MAXIMUM DEVIATIONS FROM START ===")
    for name, series in data_series.items():
        max_deviation = series.max()
        min_deviation = series.min()
        max_date = series.idxmax()
        min_date = series.idxmin()
        
        print(f"{name}:")
        print(f"  Maximum: {max_deviation:+.4f} on {max_date.strftime('%Y-%m')}")
        print(f"  Minimum: {min_deviation:+.4f} on {min_date.strftime('%Y-%m')}")
        print(f"  Range: {max_deviation - min_deviation:.4f} log points")

# Generate the normalized plot
print("Generating normalized quartiles deviation plot...")
create_normalized_quartiles_plot()
print("Normalized plot generated successfully!")
```
