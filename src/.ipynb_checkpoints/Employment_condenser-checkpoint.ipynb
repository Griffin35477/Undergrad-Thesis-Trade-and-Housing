```python
import os
import pandas as pd
import glob

# Define the folders for each year
folders = {
    "2017": "2017.q1-q4.by_area",  # Replace with your 2017 folder path
    "2018": "2018.q1-q4.by_area",  # Replace with your 2018 folder path
    "2019": "2019.q1-q4.by_area"   # Replace with your 2019 folder path
}

# Create a list to store DataFrames from all files across all years
all_dfs = []

# Process each year's folder
for year, folder_path in folders.items():
    print(f"\nProcessing folder for {year}...")
    
    # Get all CSV files in the input directory
    csv_files = glob.glob(os.path.join(folder_path, "*.csv"))
    
    # Process each CSV file
    for file in csv_files:
        print(f"  Processing file: {os.path.basename(file)}")
        
        try:
            # Read the CSV file
            df = pd.read_csv(file)
            
            # Extract the first 4 rows after the header (rows 0-3 in pandas DataFrame)
            # Create a proper copy instead of a view to avoid warnings
            first_four_data_rows = df.iloc[0:4].copy()
            
            # Add columns to identify the source
            first_four_data_rows['source_file'] = os.path.basename(file)
            first_four_data_rows['year'] = year
            first_four_data_rows['folder'] = folder_path
            
            # Append to the list of DataFrames
            all_dfs.append(first_four_data_rows)
            
        except Exception as e:
            print(f"  Error processing {os.path.basename(file)}: {str(e)}")

# Combine all DataFrames from all years into one
if all_dfs:
    combined_df = pd.concat(all_dfs, ignore_index=True)
    
    # Output file name
    output_file = "employment_data_2017_2019_combined.csv"
    
    # Save to CSV
    combined_df.to_csv(output_file, index=False)
    print(f"\nSuccessfully created {output_file} with data from all years!")
    print(f"Total number of rows: {len(combined_df)}")
else:
    print("No data was processed from any folder.")

print("\nAll processing complete!")
```
