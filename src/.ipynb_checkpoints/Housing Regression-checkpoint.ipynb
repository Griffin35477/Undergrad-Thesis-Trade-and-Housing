```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from stargazer.stargazer import Stargazer
from datetime import timedelta, datetime
from IPython.display import display, Latex, Math, HTML
from linearmodels.panel import PanelOLS
import addfips
import warnings
import os
```

```python
#Read in data files

zillow_county = pd.read_csv('../Data/zillow county.csv')
employment_levels = pd.read_csv('../Data/employment_data_2017_2019_combined.csv')
tariffs = pd.read_csv('../Data/ustariffs-by-county.csv')
population = pd.read_csv('../Data/population_data.csv')
```

```python
#Begin Editing population file, rename County column to County and drop everything besides county name and years

population.rename(columns={'Unnamed: 0': 'County'}, inplace=True)
population = population[['County','2017','2018','2019']]
```

```python
# Initialize the AddFIPS object to add FIPS codes to each row for merging purposes
af = addfips.AddFIPS()

# Create a function to extract county and state from your specific format
def add_fips_code(row):
    county_text = row['County']
    
    
    if '.' in county_text:
        # Remove the leading dot and any index
        parts = county_text.split('.')
        if len(parts) > 1:
            county_state = parts[1].strip()
        else:
            county_state = county_text
    else:
        county_state = county_text
    
    # Now split into county and state if there's a comma
    if ',' in county_state:
        county, state = county_state.split(',', 1)
        county = county.strip()
        state = state.strip()
    else:
        # Handle any special cases where format is different
        county = county_state
        # Determine state based on context or defaults
        state = None
    
    # Remove "County" from the county name as addfips usually expects just the name
    if "County" in county:
        county = county.replace(' County', '')
    
    # Get the FIPS code
    if state and county:
        fips = af.get_county_fips(county, state=state)
        return fips
    else:
        return None

# Apply the function to create a new FIPS column
population['fips'] = population.apply(add_fips_code, axis=1)
```

```python
#reset index
population_reset = population.reset_index(drop=True)

# Now melt the DataFrame to convert from wide to long format so that each row has only one value and can be matched later
population_long = pd.melt(
    population_reset,
    id_vars=['County', 'fips'],  # Columns to keep as identifiers
    value_vars=['2017', '2018', '2019'],  # Columns to unpivot
    var_name='year',  # Name for the new column containing year values
    value_name='population'  # Name for the new column containing population values
)

# Clean up the county names by removing the leading dots if needed
population_long['County'] = population_long['County'].str.replace(r'^\.\s*', '', regex=True)

# Sort by county and year for better organization
population_long = population_long.sort_values(['County', 'year'])

# Reset the index for clean display
population_long = population_long.reset_index(drop=True)

population_long['year'] = population_long['year'].astype(int)
```

```python
#Drop all columns except for identifying columns and employment levels

columns_to_keep = ['area_fips','year','qtr','area_title','month1_emplvl','month2_emplvl','month3_emplvl']
employment_levels = employment_levels[columns_to_keep]
```

```python
#Create function to create FIPS codes from the Zillow given data since given data only has state code and county code but I'm matching by 5 digit

def create_county_fips(state_fips, municipal_code):
    # Ensure state FIPS is two digits (add leading zero if needed)
    state_fips_str = str(state_fips).zfill(2)
    
    # Ensure municipal code is three digits (add leading zeros if needed)
    municipal_code_str = str(municipal_code).zfill(3)
    
    # Concatenate to form the complete county FIPS
    county_fips = state_fips_str + municipal_code_str
    
    return county_fips
```

```python
#Apply the Function to Create new FIPS column

zillow_county['CountyFIPS'] = zillow_county.apply(
    lambda row: create_county_fips(row['StateCodeFIPS'], row['MunicipalCodeFIPS']), 
    axis=1
)
```

```python
#drop all data points thats not county in zillow data

zillow_county = zillow_county[zillow_county['RegionType'] == 'county']
```

```python
#Clean up Zillow county Data 

columns_to_drop = ['RegionID','SizeRank', 'RegionType','StateName','State','StateCodeFIPS','MunicipalCodeFIPS','Metro']
zillow_county = zillow_county.drop(columns = columns_to_drop)
```

```python
#Rearrange Zillow Data for Regression and Merge
# Melt the date columns into rows
melted_df = pd.melt(
    zillow_county,
    id_vars=['RegionName', 'CountyFIPS'],  # Keep Metro and county_fips as identifiers
    var_name='Date',                   # Column names (dates) become values in the 'Date' column
    value_name='HousingPrice'          # Values become housing prices
)

# Convert the date column to datetime
melted_df['Date'] = pd.to_datetime(melted_df['Date'])

# Sort by date and metro
melted_df = melted_df.sort_values(['Date', 'RegionName'])

# Reset the index
melted_df = melted_df.reset_index(drop=True)

```

```python
#Convert Date column to Datetime format
melted_df['Date'] = pd.to_datetime(melted_df['Date'])

# Filter to keep only data from 2017-2019
melted_df = melted_df[(melted_df['Date'] >= '2017-01-01') & (melted_df['Date'] <= '2019-12-31')]

# Reset index after filtering
melted_df = melted_df.reset_index(drop=True)
```

```python
#clean tariffs data frame by dropping everything but identifiyng factors and tariff levels 
tariffs = tariffs[['time','area_fips','tariff']]

# First, convert the 'time' column to datetime
tariffs['time'] = pd.to_datetime(tariffs['time'])

# Shift dates back by one day
tariffs['time'] = tariffs['time'] - timedelta(days=1)

# Filter to keep only dates between 2017-01-01 and 2019-12-31
filtered_tariffs = tariffs[(tariffs['time'] >= '2017-01-01') & (tariffs['time'] <= '2019-12-31')]

# Reset the index
filtered_tariffs = filtered_tariffs.reset_index(drop=True)

filtered_tariffs['area_fips'] = filtered_tariffs['area_fips'].astype(str)
```

```python
#Drop all employment data points that aren't that of a county
county_employment_levels = employment_levels[employment_levels['area_title'].str.contains('County', case=False, na=False)]

# Create an empty list to store our transformed data
transformed_data = []

# Iterate through each row in the filtered DataFrame
for _, row in county_employment_levels.iterrows():
    area_fips = row['area_fips']
    area_title = row['area_title']
    year = row['year']
    quarter = row['qtr']
    
    # Calculate the starting month for each quarter
    base_month = (quarter - 1) * 3 + 1
    
    # For each of the three months in the quarter
    for month_idx in range(3):
        # Calculate the actual month number
        month_num = base_month + month_idx
        
        # Create the date (last day of month)
        if month_num == 2:  # February
            if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):  # Leap year
                day = 29
            else:
                day = 28
        elif month_num in [4, 6, 9, 11]:  # 30-day months
            day = 30
        else:  # 31-day months
            day = 31
        
        # Create date string in the format 'YYYY-MM-DD'
        date_str = f"{year}-{month_num:02d}-{day}"
        
        # Get the employment level for this month
        employment_level = row[f'month{month_idx+1}_emplvl']
        
        # Add to our transformed data
        transformed_data.append({
            'CountyFIPS': area_fips,
            'RegionName': area_title,
            'Date': date_str,
            'EmploymentLevel': employment_level
        })

# Create a new DataFrame from our transformed data
transformed_employment_df = pd.DataFrame(transformed_data)

# Convert Date to datetime
transformed_employment_df['Date'] = pd.to_datetime(transformed_employment_df['Date'])

# Sort by CountyFIPS and Date
transformed_employment_df = transformed_employment_df.sort_values(['CountyFIPS', 'Date']).reset_index(drop=True)
```

```python
# Merge the transformed employment data with housing price data
merged_df = pd.merge(
    transformed_employment_df,  # Employment data
    melted_df,        # Housing price data
    how='inner',               # Keep only matching rows
    left_on=['CountyFIPS', 'Date'],
    right_on=['CountyFIPS', 'Date']
)
```

```python
#merge again to add tariffs
merged_df = pd.merge(
    merged_df,
    filtered_tariffs,
    how = 'inner',
    left_on=['CountyFIPS', 'Date'],
    right_on = ['area_fips','time']
)
```

```python
# Extract year from the Date column
merged_df['year'] = merged_df['Date'].dt.year
#merge one last time to add population
merged_df = pd.merge(
    merged_df,
    population_long,
    how = 'inner',
    left_on = ['CountyFIPS','year'],
    right_on = ['fips','year']
)
```

```python
#After confirmation drop Region names
merged_df = merged_df.drop(columns = ['RegionName_x','RegionName_y','area_fips','year','time','County','fips'])

# Drop rows with NaN values in any column
merged_df = merged_df.dropna()

# Reset the index after dropping rows
merged_df = merged_df.reset_index(drop=True)

# Remove commas from population column first, then convert to integer
merged_df['population'] = merged_df['population'].astype(str).str.replace(',', '')

# Now convert to integer
merged_df['population'] = pd.to_numeric(merged_df['population'], errors='coerce')
```

```python
#make dummies for fixed effect regression later

county_dummies = pd.get_dummies(merged_df['CountyFIPS'], prefix = 'County')
merged_data_county_dummies = pd.concat([merged_df,county_dummies],axis = 1)
merged_data_county_dummies = merged_data_county_dummies.drop(columns = ['Date','CountyFIPS','HousingPrice','population'])


#do same thing with dates 
date_dummies = pd.get_dummies(merged_df['Date'],prefix = 'Date')
merged_data_date_dummies = pd.concat([merged_df,date_dummies],axis = 1)
merged_data_date_dummies = merged_data_date_dummies.drop(columns = ['Date','CountyFIPS','HousingPrice','population'])

#create final data set with all dummies for last step of regression
merged_data_large = pd.concat([merged_data_county_dummies,date_dummies],axis = 1)
#merged_data_large = merged_data_large.drop(columns = ['Date','CountyFIPS','HousingPrice'])
```

```python
merged_df['HousingPrice'] = np.log(merged_df['HousingPrice'])
```

```python

warnings.filterwarnings('ignore')

def create_12month_differences(df):
    """
    Calculate 12-month differences for all variables
    All variables will be 12-month changes for consistency
    """
    # Make a copy to avoid modifying original data
    df = df.copy()
    
    # Ensure Date column is datetime
    df['Date'] = pd.to_datetime(df['Date'])
    
    # Sort by county and date
    df = df.sort_values(['CountyFIPS', 'Date'])
    
    # Calculate 12-month lagged values by county
    df['HousingPrice_12m'] = df.groupby('CountyFIPS')['HousingPrice'].shift(12)
    df['EmploymentLevel_12m'] = df.groupby('CountyFIPS')['EmploymentLevel'].shift(12)
    df['tariff_12m'] = df.groupby('CountyFIPS')['tariff'].shift(12)
    df['population_12m'] = df.groupby('CountyFIPS')['population'].shift(12)
    
    # Calculate 12-month differences for all variables
    # Housing prices (already in log form)
    df['HousingPrice_diff'] = df['HousingPrice'] - df['HousingPrice_12m']
    
    # Employment: calculate log difference
    # First ensure we have positive employment levels
    df = df[(df['EmploymentLevel'] > 0) & (df['EmploymentLevel_12m'] > 0)]
    df['EmploymentLevel_diff'] = np.log(df['EmploymentLevel']) - np.log(df['EmploymentLevel_12m'])
    
    # Tariff: calculate difference (since tariffs are already in percentage/ratio form)
    df['tariff_diff'] = df['tariff'] - df['tariff_12m']
    
    # Use the earlier year's population as weight (the lagged population)
    df['weight_population'] = df['population_12m']
    
    # Drop rows with missing 12-month differences
    df = df.dropna(subset=['HousingPrice_diff', 'EmploymentLevel_diff', 'tariff_diff', 'weight_population'])
    
    # Keep only the columns we need
    result_df = df[['CountyFIPS', 'Date', 'HousingPrice_diff', 'tariff_diff', 
                   'EmploymentLevel_diff', 'weight_population']].copy()
    
    # Rename for consistency with previous code
    result_df = result_df.rename(columns={
        'HousingPrice_diff': 'HousingPrice',
        'tariff_diff': 'tariff',
        'EmploymentLevel_diff': 'EmploymentLevel',
        'weight_population': 'population'
    })
    
    print(f"Analysis period: {result_df['Date'].min()} to {result_df['Date'].max()}")
    print(f"Panel dataset has {len(result_df)} observations from {result_df['CountyFIPS'].nunique()} counties")
    print(f"Average observations per county: {len(result_df) / result_df['CountyFIPS'].nunique():.1f}")
    print(f"Average 12-month log housing price change: {result_df['HousingPrice'].mean():.4f}")
    print(f"Average 12-month log employment change: {result_df['EmploymentLevel'].mean():.4f}")
    print(f"Average 12-month tariff change: {result_df['tariff'].mean():.4f}")
    print(f"Standard deviation of 12-month log price change: {result_df['HousingPrice'].std():.4f}")
    print(f"Standard deviation of 12-month log employment change: {result_df['EmploymentLevel'].std():.4f}")
    print(f"Standard deviation of 12-month tariff change: {result_df['tariff'].std():.4f}")
    
    return result_df

def create_dummy_variables(df):
    """
    Create county and date dummy variables for fixed effects
    """
    # County dummies
    county_dummies = pd.get_dummies(df['CountyFIPS'], prefix='County')
    merged_data_county_dummies = pd.concat([df, county_dummies], axis=1)
    merged_data_county_dummies = merged_data_county_dummies.drop(columns=['Date', 'CountyFIPS', 'HousingPrice', 'population'])
    
    # Date dummies  
    date_dummies = pd.get_dummies(df['Date'], prefix='Date')
    merged_data_date_dummies = pd.concat([df, date_dummies], axis=1)
    merged_data_date_dummies = merged_data_date_dummies.drop(columns=['Date', 'CountyFIPS', 'HousingPrice', 'population'])
    
    # Combined dummies
    merged_data_large = pd.concat([merged_data_county_dummies, date_dummies], axis=1)
    
    return merged_data_county_dummies, merged_data_date_dummies, merged_data_large

def add_stars(coef, pval):
    """Add significance stars to coefficients"""
    if pval < 0.01:
        return f"{coef:.3f}***"
    elif pval < 0.05:
        return f"{coef:.3f}**"
    elif pval < 0.1:
        return f"{coef:.3f}*"
    else:
        return f"{coef:.3f}"

def create_regression_table(reg_list):
    """Create a formatted regression table matching your original format"""
    # Create an empty dataframe for the table
    columns = ['(1)', '(2)', '(3)', '(4)', '(5)', '(6)']
    rows = ['Δ Tariff', '', 'Δ Log Employment', '', 'Fixed Effects:', 'County', 'Time', 'Observations', 'R-squared']
    table = pd.DataFrame(index=rows, columns=columns)
    
    # Extract coefficient values with stars and standard errors
    # Tariff coefficients
    for i, reg in enumerate(reg_list):
        col = columns[i]
        table.loc['Δ Tariff', col] = add_stars(reg.params['tariff'], reg.pvalues['tariff'])
        table.loc['', col] = f"({reg.bse['tariff']:.3f})"
    
    # Employment coefficients (for regressions that include it)
    employment_regs = [reg_list[3], reg_list[5]]  # reg_4 and reg_6
    employment_cols = [columns[3], columns[5]]
    
    for reg, col in zip(employment_regs, employment_cols):
        if 'EmploymentLevel' in reg.params:
            table.loc['Δ Log Employment', col] = add_stars(reg.params['EmploymentLevel'], reg.pvalues['EmploymentLevel'])
            table.loc['', col] = f"({reg.bse['EmploymentLevel']:.3f})"
    
    # Fixed effects indicators
    table.loc['Fixed Effects:'] = ""
    table.loc['County'] = ['N', 'N', 'N', 'N', 'Y', 'Y']
    table.loc['Time'] = ['N', 'N', 'Y', 'Y', 'Y', 'Y']
    
    # Model statistics
    table.loc['Observations'] = [f"{int(reg.nobs)}" for reg in reg_list]
    table.loc['R-squared'] = [f"{reg.rsquared:.3f}" for reg in reg_list]
    
    # Fill NaN values with empty strings
    table = table.fillna("")
    
    return table

# Apply the analysis to your merged data
print("Creating 12-month differences panel dataset (all variables)...")
panel_data = create_12month_differences(merged_df)

# Create dummy variables for fixed effects
merged_data_county_dummies, merged_data_date_dummies, merged_data_large = create_dummy_variables(panel_data)

# Run the same 6 regressions as in your original model
print("Running panel regressions...")

# Regression 1: Simple OLS with just tariff change
reg_1 = sm.OLS(
    endog=panel_data['HousingPrice'],
    exog=sm.add_constant(panel_data['tariff'])
).fit()

# Regression 2: WLS with population weights (using earlier year's population)
reg_2 = sm.WLS(
    endog=panel_data['HousingPrice'],
    exog=sm.add_constant(panel_data['tariff']),
    weights=panel_data['population']
).fit()

# Regression 3: WLS with time dummies (but no employment change)
reg_3 = sm.WLS(
    endog=panel_data['HousingPrice'],
    exog=sm.add_constant(merged_data_date_dummies.drop(columns=['EmploymentLevel'])),
    weights=panel_data['population']
).fit()

# Regression 4: WLS with time dummies and employment change
reg_4 = sm.WLS(
    endog=panel_data['HousingPrice'],
    exog=sm.add_constant(merged_data_date_dummies),
    weights=panel_data['population']
).fit()

# Regression 5: WLS with county and time dummies (but no employment change)
reg_5 = sm.WLS(
    endog=panel_data['HousingPrice'],
    exog=sm.add_constant(merged_data_large.drop(columns=['EmploymentLevel'])),
    weights=panel_data['population']
).fit()

# Regression 6: WLS with county and time dummies and employment change
reg_6 = sm.WLS(
    endog=panel_data['HousingPrice'],
    exog=sm.add_constant(merged_data_large),
    weights=panel_data['population']
).fit()

# Create and display the table
table = create_regression_table([reg_1, reg_2, reg_3, reg_4, reg_5, reg_6])

# Convert the table to HTML for better display
html = f"""
<h3>Effect of Tariff Changes on 12-Month Housing Price Changes (2017-2019)</h3>
{table.to_html()}
<p><small>Notes: *** p<0.01, ** p<0.05, * p<0.1. Standard errors shown in 
parentheses. Data covers 2018-2019 by county (12-month differences starting from 2017). 
The dependent variable is the 12-month log difference in county-level housing prices. 
Δ Tariff is the 12-month change in tariff exposure. Δ Log Employment is the 12-month 
change in log employment levels. All regressions use population weights from the earlier year.</small></p>
"""

display(HTML(html))

# Generate LaTeX output
latex_table = table.to_latex(caption="Effect of Tariff Changes on 12-Month Housing Price Changes", label="tab:tariffs_changes")

# Save to file with notes
with open('regression_table_all_changes.tex', 'w') as f:
    f.write("\\begin{table}[htbp]\n")
    f.write("\\centering\n")
    f.write("\\caption{Effect of Tariff Changes on 12-Month Housing Price Changes (2017-2019)}\n")
    f.write("\\label{tab:tariffs_changes}\n")
    f.write(table.to_latex(index=True))
    f.write("\\begin{tablenotes}\n")
    f.write("\\small\n")
    f.write("\\item Notes: *** p<0.01, ** p<0.05, * p<0.1. Standard errors shown in\n")
    f.write("parentheses. Data covers 2018-2019 by county (12-month differences starting from 2017).\n")
    f.write("The dependent variable is the 12-month log difference in county-level housing prices.\n")
    f.write("$\\Delta$ Tariff is the 12-month change in tariff exposure. $\\Delta$ Log Employment is\n")
    f.write("the 12-month change in log employment levels. All regressions use population weights\n")
    f.write("from the earlier year.\n")
    f.write("\\end{tablenotes}\n")
    f.write("\\end{table}\n")

# Print the LaTeX table to copy directly
print("\n" + "="*60)
print("LATEX TABLE CODE")
print("="*60)
print(latex_table)

print("\n" + "="*60)
print("LaTeX table saved to: regression_table_all_changes.tex")
print("="*60)

# Print summary statistics
print("\n" + "="*60)
print("SUMMARY STATISTICS")
print("="*60)
print(f"Number of county-month observations: {len(panel_data)}")
print(f"Number of unique counties: {panel_data['CountyFIPS'].nunique()}")
print(f"Average observations per county: {len(panel_data) / panel_data['CountyFIPS'].nunique():.1f}")
print(f"Average 12-month tariff change: {panel_data['tariff'].mean():.4f}")
print(f"Average 12-month log price change: {panel_data['HousingPrice'].mean():.4f}")
print(f"Average 12-month log employment change: {panel_data['EmploymentLevel'].mean():.4f}")
print(f"Std dev of 12-month tariff change: {panel_data['tariff'].std():.4f}")
print(f"This corresponds to an average 12-month price change of: {(np.exp(panel_data['HousingPrice'].mean()) - 1) * 100:.2f}%")
print(f"This corresponds to an average 12-month employment change of: {(np.exp(panel_data['EmploymentLevel'].mean()) - 1) * 100:.2f}%")

# Print key regression results
print("\n" + "="*60)
print("KEY RESULTS")
print("="*60)
print(f"Tariff change coefficient (simple OLS): {reg_1.params['tariff']:.4f} (p-value: {reg_1.pvalues['tariff']:.4f})")
print(f"Tariff change coefficient (weighted): {reg_2.params['tariff']:.4f} (p-value: {reg_2.pvalues['tariff']:.4f})")
print(f"Tariff change coefficient (time FE + controls): {reg_4.params['tariff']:.4f} (p-value: {reg_4.pvalues['tariff']:.4f})")
print(f"Tariff change coefficient (county + time FE + controls): {reg_6.params['tariff']:.4f} (p-value: {reg_6.pvalues['tariff']:.4f})")

# Interpretation for the most complete specification
print("\n" + "="*60)
print("INTERPRETATION (Full Model with County + Time FE)")
print("="*60)
tariff_coef = reg_6.params['tariff']
if tariff_coef < 0:
    print(f"A 1 percentage point increase in 12-month tariff change is associated with")
    print(f"a {abs(tariff_coef)*100:.2f} percentage point decrease in 12-month housing price growth")
    print(f"(approximately {(1 - np.exp(tariff_coef))*100:.2f}% change in housing prices)")
else:
    print(f"A 1 percentage point increase in 12-month tariff change is associated with")
    print(f"a {tariff_coef*100:.2f} percentage point increase in 12-month housing price growth")
    print(f"(approximately {(np.exp(tariff_coef) - 1)*100:.2f}% change in housing prices)")

print("\n" + "="*60)
print("MODEL SPECIFICATION")
print("="*60)
print("All variables are now 12-month changes:")
print("- Dependent variable: 12-month change in log housing prices")
print("- Independent variable: 12-month change in tariff exposure")
print("- Control variable: 12-month change in log employment")
print("- Weights: Population from the earlier year (baseline year)")
print("This specification examines how changes in tariff exposure affect changes in housing prices.")
```

```python
warnings.filterwarnings('ignore')

# Set style for plots
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")

def calc_12month_diff(df):
    """Calculate 12-month differences in log housing prices"""
    # Make a copy to avoid modifying original data
    df = df.copy()
    
    # Ensure Date column is datetime
    df['Date'] = pd.to_datetime(df['Date'])
    
    # Sort by county and date
    df = df.sort_values(['CountyFIPS', 'Date'])
    
    # Use the housing prices directly (assuming they're already in log form)
    # Rename for clarity - these are log housing prices
    df['log_hprices'] = df['HousingPrice']
    
    # Calculate 12-month lagged log prices by county
    df['log_hprices_12m'] = df.groupby('CountyFIPS')['log_hprices'].shift(12)
    
    # Calculate 12-month log difference
    df['hprices_12m_diff'] = df['log_hprices'] - df['log_hprices_12m']
    
    # Drop rows with missing 12-month differences
    df = df.dropna(subset=['hprices_12m_diff'])
    
    return df

def run_regression_for_month(data, month):
    """Run panel regression for a specific month"""
    try:
        # Filter data for specific month
        month_data = data[data['Date'] == month].copy()
        
        if len(month_data) < 10:  # Need minimum observations
            return None
        
        # Set up panel data structure
        month_data = month_data.set_index(['CountyFIPS', 'Date'])
        
        # Check if we have variation in both dependent and independent variables
        if month_data['hprices_12m_diff'].std() == 0 or month_data['tariff'].std() == 0:
            return None
        
        # Run weighted panel regression with entity fixed effects
        model = PanelOLS(
            dependent=month_data['hprices_12m_diff'],
            exog=month_data[['tariff']],
            weights=month_data['population'],
            entity_effects=True,
            time_effects=False
        )
        
        results = model.fit(cov_type='clustered', cluster_entity=True)
        
        return {
            'coefficient': results.params['tariff'],
            'std_error': results.std_errors['tariff'],
            'time_period': month,
            'observations': len(month_data),
            'r_squared': results.rsquared
        }
    except Exception as e:
        print(f"Error processing month {month}: {str(e)}")
        return None

# Apply the 12-month difference calculation to your merged data
print("Calculating 12-month log differences...")
panel_data = calc_12month_diff(merged_df)

print(f"Panel data shape after calculating differences: {panel_data.shape}")
print(f"Date range: {panel_data['Date'].min()} to {panel_data['Date'].max()}")

# Get unique months for regression (only those with 12-month lags available)
unique_months = sorted(panel_data['Date'].unique())
print(f"Running regressions for {len(unique_months)} months...")

# Run regression for each month
results_list = []
for i, month in enumerate(unique_months):
    if i % 6 == 0:  # Progress indicator every 6 months
        # Convert numpy datetime64 to pandas timestamp for formatting
        month_pd = pd.to_datetime(month)
        print(f"Processing month {i+1}/{len(unique_months)}: {month_pd.strftime('%Y-%m')}")
    
    result = run_regression_for_month(panel_data, month)
    if result is not None:
        results_list.append(result)

# Convert results to DataFrame
results_df = pd.DataFrame(results_list)
print(f"Successfully estimated coefficients for {len(results_df)} months")

# Create the time series plot
plt.figure(figsize=(12, 8))

# Extract data for plotting
dates = pd.to_datetime(results_df['time_period'])
coefficients = results_df['coefficient']
std_errors = results_df['std_error']

# Calculate confidence intervals
ci_upper = coefficients + 1.96 * std_errors
ci_lower = coefficients - 1.96 * std_errors

# Plot the coefficient line
plt.plot(dates, coefficients, 'b--o', linewidth=2, markersize=6, 
         label='Tariff Effect', markerfacecolor='blue', markeredgecolor='darkblue')

# Add confidence interval shading
plt.fill_between(dates, ci_lower, ci_upper, alpha=0.2, color='blue', 
                label='95% Confidence Interval')

# Add horizontal line at zero
plt.axhline(y=0, color='red', linestyle='--', alpha=0.7, linewidth=1.5, 
           label='Zero Effect')

# Add vertical line for March 2018 (First wave of tariffs)
march_2018 = pd.to_datetime('2018-03-01')
plt.axvline(x=march_2018, color='orange', linestyle='-', alpha=0.8, linewidth=2, 
           label='First Wave of Tariffs (March 2018)')

# Customize the plot
plt.title('Estimates of Tariff Effect on Housing Prices Over Time', fontsize=14, fontweight='bold')
plt.xlabel('Time')
plt.ylabel('Estimated Tariff Effect')

# Format x-axis
plt.xticks(rotation=45)
plt.gca().xaxis.set_major_locator(plt.MaxNLocator(12))

# Add grid
plt.grid(True, alpha=0.3, linestyle='-', linewidth=0.5)

# Add legend
plt.legend(loc='best', fontsize=11, framealpha=0.9)

# Adjust layout
plt.tight_layout()

# Create output path in Images folder
os.makedirs('Images', exist_ok=True)
output_path = os.path.join('Images', 'est_tariff_effct.png')

# Save the plot with high resolution
plt.savefig(output_path, 
            dpi=300,
            bbox_inches='tight',
            pad_inches=0.1)
plt.show()
plt.close()

print(f"Plot saved to: {output_path}")
```
